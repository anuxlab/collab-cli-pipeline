## Reusable definitions

Define functions and classes that are used by later notebook sections.

```python
# Generate synthetic time series data
def generate_time_series(n_days=180):
    # Trend component
    trend = np.linspace(0, 10, n_days)
    
    # Weekly seasonality component
    seasonality = 10 * np.sin(np.linspace(0, 2 * np.pi * 26, n_days))
    
    # Noise component
    noise = np.random.normal(0, 2, n_days)
    
    # Combine components
    series = trend + seasonality + noise
    return series
```

## Experiment configuration

Define reproducible parameters, paths, seeds and settings.

```python
# Generate the time series
series = generate_time_series()

# Create a DataFrame
df = pd.DataFrame({'date': pd.date_range(start='2023-01-01', periods=180, freq='D'),
                   'value': series})

# Train-test split
train_size = int(0.8 * len(df))
train_df = df.iloc[:train_size]
test_df = df.iloc[train_size:]
```

## Data validation and quality checks

Check schema, ranges, duplicates, missing values and input assumptions.

## Reusable definitions

Define functions and classes that are used by later notebook sections.

```python
# Feature engineering
def create_features(df):
    df['lag_1'] = df['value'].shift(1)
    df['lag_7'] = df['value'].shift(7)
    df['rolling_mean_7'] = df['value'].rolling(window=7).mean()
    df['rolling_std_7'] = df['value'].rolling(window=7).std()
    return df.dropna()
```

## Experiment configuration

Define reproducible parameters, paths, seeds and settings.

```python
train_df = create_features(train_df)
test_df = create_features(test_df)
```

## Reusable definitions

Define functions and classes that are used by later notebook sections.

```python
# Baseline models
def naive_forecast(train_df, test_df):
    return train_df['value'].iloc[-1] * np.ones(len(test_df))

def seasonal_naive_forecast(train_df, test_df):
    return train_df['value'].iloc[-7:].values
```

## Forecasting and temporal baselines

Establish naive or seasonal forecasts and fit time-aware prediction methods.
