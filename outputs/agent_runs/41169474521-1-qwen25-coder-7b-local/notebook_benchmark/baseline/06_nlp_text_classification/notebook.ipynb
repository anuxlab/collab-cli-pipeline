## Environment and imports

Import dependencies and prepare the notebook runtime.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns
```

## Experiment execution

Run the remaining experiment logic in its original order.

```python
# Set random seed for reproducibility
np.random.seed(42)

# Create synthetic dataset
np.random.seed(42)
```

## Experiment configuration

Define reproducible parameters, paths, seeds and settings.

```python
positive_texts = ["I love this product!", "This is the best!", "Absolutely fantastic!"]
negative_texts = ["I hate this product!", "This is terrible!", "Absolutely awful!"]
texts = positive_texts * 10 + negative_texts * 10
labels = [1] * len(positive_texts) + [0] * len(negative_texts)

# Shuffle the dataset
indices = np.arange(len(texts))
```

## Experiment execution

Run the remaining experiment logic in its original order.

```python
np.random.shuffle(indices)
```

## Experiment configuration

Define reproducible parameters, paths, seeds and settings.

```python
texts = [texts[i] for i in indices]
labels = [labels[i] for i in indices]
```

## Train-test split and cross-validation

Create evaluation partitions without leaking information across splits.

```python
# Split the dataset into training and test sets
X_train, X_test, y_train, y_test = train_test_split(texts, labels, test_size=0.2, random_state=42, stratify=labels)
```

## Text preprocessing and vectorization

Transform raw text into model-ready representations while preserving the train/test boundary.

```python
# Create a pipeline with TfidfVectorizer and LogisticRegression
pipeline = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('logreg', LogisticRegression(random_state=42))
])
```

## Feature engineering and preprocessing

Clean, transform and prepare features for downstream analysis.

```python
# Train the pipeline
pipeline.fit(X_train, y_train)

# Make predictions
y_pred = pipeline.predict(X_test)
```

## Text classification and error analysis

Fit text classifiers, evaluate class-wise behaviour and inspect misclassified examples.

```python
# Evaluate the model
report = classification_report(y_test, y_pred)
```

## Evaluation and metrics

Measure model or experiment performance using appropriate metrics.

```python
conf_matrix = confusion_matrix(y_test, y_pred)
accuracy = accuracy_score(y_test, y_pred)
```
